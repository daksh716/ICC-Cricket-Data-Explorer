```python
print("=" * 60)
print("STEP 4 - ODI PLAYER ANALYSIS")
print("=" * 60)

player_stats = (
    odi_deliveries
    .groupby("Batter")
    .agg(
        Runs=("Runs by Batter", "sum"),
        Balls=("Runs by Batter", "count"),
        Matches=("Match ID", "nunique"),
        Fours=("Runs by Batter", lambda x: (x == 4).sum()),
        Sixes=("Runs by Batter", lambda x: (x == 6).sum())
    )
)

player_stats["Strike Rate"] = (
    player_stats["Runs"] / player_stats["Balls"] * 100
).round(2)

player_stats["Average"] = (
    odi_deliveries
    .groupby("Batter")["Runs by Batter"]
    .sum() /
    odi_deliveries
    .groupby("Batter")["Match ID"].nunique()
).round(2)

player_stats["Highest Score"] = (
    odi_deliveries
    .groupby(["Match ID", "Batter"])["Runs by Batter"]
    .sum()
    .groupby("Batter")
    .max()
)

player_stats["50s"] = (
    odi_deliveries
    .groupby(["Match ID", "Batter"])["Runs by Batter"]
    .sum()
    .groupby("Batter")
    .apply(lambda x: ((x >= 50) & (x < 100)).sum())
)

player_stats["100s"] = (
    odi_deliveries
    .groupby(["Match ID", "Batter"])["Runs by Batter"]
    .sum()
    .groupby("Batter")
    .apply(lambda x: (x >= 100).sum())
)

player_stats = player_stats.sort_values(
    "Runs",
    ascending=False
)

print("\nTOP 20 ODI PLAYERS BY RUNS")
print("-" * 40)
print(player_stats.head(20))

print("\nPLAYER STATISTICS COLUMNS")
print("-" * 40)
print(player_stats.columns.tolist())

print("\nTOTAL PLAYERS ANALYZED:",
      len(player_stats))

print("\nSTEP 4 COMPLETED")
```

```python
print("=" * 60)
print("STEP 5 - ODI BOWLER ANALYSIS")
print("=" * 60)

bowler_stats = (
    odi_deliveries
    .groupby("Bowler")
    .agg(
        Runs_Conceded=("Total Runs", "sum"),
        Balls=("Total Runs", "count"),
        Matches=("Match ID", "nunique"),
        Wickets=("Player Out", lambda x: x.notna().sum())
    )
)

bowler_stats["Overs"] = (
    bowler_stats["Balls"] / 6
).round(2)

bowler_stats["Economy"] = (
    bowler_stats["Runs_Conceded"] /
    bowler_stats["Balls"] * 6
).round(2)

bowler_match_wickets = (
    odi_deliveries[
        odi_deliveries["Player Out"].notna()
    ]
    .groupby(["Match ID", "Bowler"])["Player Out"]
    .count()
)

bowler_stats["4 Wickets"] = (
    bowler_match_wickets
    .groupby("Bowler")
    .apply(lambda x: (x == 4).sum())
)

bowler_stats["5 Wickets"] = (
    bowler_match_wickets
    .groupby("Bowler")
    .apply(lambda x: (x >= 5).sum())
)

bowler_stats = bowler_stats.sort_values(
    "Wickets",
    ascending=False
)

print("\nTOP 20 ODI BOWLERS BY WICKETS")
print("-" * 40)
print(bowler_stats.head(20))

print("\nBOWLER STATISTICS COLUMNS")
print("-" * 40)
print(bowler_stats.columns.tolist())

print("\nTOTAL BOWLERS ANALYZED:",
      len(bowler_stats))

print("\nSTEP 5 COMPLETED")
```

```python
print("=" * 60)
print("STEP 6 - ODI TEAM ANALYSIS")
print("=" * 60)

team_list = pd.unique(
    pd.concat([
        odi_matches["Team 1"],
        odi_matches["Team 2"]
    ])
)

team_stats = []

for team in team_list:

    matches = odi_matches[
        (odi_matches["Team 1"] == team) |
        (odi_matches["Team 2"] == team)
    ]

    wins = (
        matches["Winner"] == team
    ).sum()

    losses = (
        (matches["Winner"].notna()) &
        (matches["Winner"] != team)
    ).sum()

    matches_played = len(matches)

    win_percentage = (
        wins / matches_played * 100
        if matches_played > 0 else 0
    )

    runs_scored = odi_deliveries[
        odi_deliveries["Inning Team"] == team
    ]["Total Runs"].sum()

    team_stats.append([
        team,
        matches_played,
        wins,
        losses,
        round(win_percentage, 2),
        runs_scored
    ])

team_stats = pd.DataFrame(
    team_stats,
    columns=[
        "Team",
        "Matches",
        "Wins",
        "Losses",
        "Win %",
        "Runs Scored"
    ]
)

team_stats = team_stats.sort_values(
    "Wins",
    ascending=False
)

print("\nTOP ODI TEAMS BY WINS")
print("-" * 40)
print(team_stats.head(20).to_string(index=False))

print("\nTOP ODI TEAMS BY WIN %")
print("-" * 40)

print(
    team_stats[
        team_stats["Matches"] >= 50
    ]
    .sort_values("Win %", ascending=False)
    .head(15)
    .to_string(index=False)
)

print("\nTOP ODI TEAMS BY RUNS SCORED")
print("-" * 40)

print(
    team_stats
    .sort_values("Runs Scored", ascending=False)
    .head(15)
    .to_string(index=False)
)

print("\nTOTAL TEAMS ANALYZED:",
      len(team_stats))

print("\nSTEP 6 COMPLETED")
```

```python
print("=" * 60)
print("STEP 7 - T20I PLAYER ANALYSIS")
print("=" * 60)

t20_player_stats = (
    t20_deliveries
    .groupby("Batter")
    .agg(
        Runs=("Batter Runs", "sum"),
        Matches=("Match ID", "nunique"),
        Fours=("Batter Runs", lambda x: (x == 4).sum()),
        Sixes=("Batter Runs", lambda x: (x == 6).sum())
    )
)

t20_player_balls = (
    t20_deliveries
    .groupby(["Match ID", "Batter"])["Batter Balls Faced"]
    .max()
    .groupby("Batter")
    .sum()
)

t20_player_stats["Balls"] = t20_player_balls

t20_player_stats["Strike Rate"] = (
    t20_player_stats["Runs"] /
    t20_player_stats["Balls"] * 100
).round(2)

t20_match_runs = (
    t20_deliveries
    .groupby(["Match ID", "Batter"])["Batter Runs"]
    .sum()
)

t20_player_stats["Highest Score"] = (
    t20_match_runs
    .groupby("Batter")
    .max()
)

t20_player_stats["50s"] = (
    t20_match_runs
    .groupby("Batter")
    .apply(lambda x: ((x >= 50) & (x < 100)).sum())
)

t20_player_stats["100s"] = (
    t20_match_runs
    .groupby("Batter")
    .apply(lambda x: (x >= 100).sum())
)

t20_dismissals = (
    t20_deliveries
    .groupby("Batter")["Player Out"]
    .apply(lambda x: x.notna().sum())
)

t20_player_stats["Average"] = (
    t20_player_stats["Runs"] /
    t20_dismissals.replace(0, np.nan)
).round(2)

t20_player_stats = t20_player_stats[
    [
        "Runs",
        "Balls",
        "Matches",
        "Fours",
        "Sixes",
        "Strike Rate",
        "Average",
        "Highest Score",
        "50s",
        "100s"
    ]
]

t20_player_stats = t20_player_stats.sort_values(
    "Runs",
    ascending=False
)

print("\nTOP 20 T20I PLAYERS BY RUNS")
print("-" * 40)
print(t20_player_stats.head(20).to_string())

print("\nTOTAL T20I PLAYERS ANALYZED:",
      len(t20_player_stats))

print("\nSTEP 7 COMPLETED")
```

```python
print("=" * 60)
print("STEP 8 - T20I BOWLER ANALYSIS")
print("=" * 60)

t20_bowler_stats = (
    t20_deliveries
    .groupby("Bowler")
    .agg(
        Runs_Conceded=("Bowler Runs Conceded", "sum"),
        Matches=("Match ID", "nunique"),
        Wickets=("Player Out", lambda x: x.notna().sum())
    )
)

t20_bowler_balls = (
    t20_deliveries
    .groupby(["Match ID", "Bowler"])["Valid Ball"]
    .sum()
    .groupby("Bowler")
    .sum()
)

t20_bowler_stats["Balls"] = t20_bowler_balls

t20_bowler_stats["Overs"] = (
    t20_bowler_stats["Balls"] / 6
).round(2)

t20_bowler_stats["Economy"] = (
    t20_bowler_stats["Runs_Conceded"] /
    t20_bowler_stats["Balls"] * 6
).round(2)

t20_match_wickets = (
    t20_deliveries[
        t20_deliveries["Player Out"].notna()
    ]
    .groupby(["Match ID", "Bowler"])["Player Out"]
    .count()
)

t20_bowler_stats["4 Wickets"] = (
    t20_match_wickets
    .groupby("Bowler")
    .apply(lambda x: (x == 4).sum())
)

t20_bowler_stats["5 Wickets"] = (
    t20_match_wickets
    .groupby("Bowler")
    .apply(lambda x: (x >= 5).sum())
)

t20_bowler_stats = t20_bowler_stats[
    [
        "Runs_Conceded",
        "Balls",
        "Matches",
        "Wickets",
        "Overs",
        "Economy",
        "4 Wickets",
        "5 Wickets"
    ]
]

t20_bowler_stats = t20_bowler_stats.sort_values(
    "Wickets",
    ascending=False
)

print("\nTOP 20 T20I BOWLERS BY WICKETS")
print("-" * 40)
print(t20_bowler_stats.head(20).to_string())

print("\nBOWLER STATISTICS COLUMNS")
print("-" * 40)
print(t20_bowler_stats.columns.tolist())

print("\nTOTAL T20I BOWLERS ANALYZED:",
      len(t20_bowler_stats))

print("\nSTEP 8 COMPLETED")
```

```python
print("=" * 60)
print("STEP 9 - T20I TEAM ANALYSIS")
print("=" * 60)

t20_team_list = pd.unique(
    pd.concat([
        t20_deliveries["Bat First"],
        t20_deliveries["Bat Second"]
    ])
)

t20_team_stats = []

for team in t20_team_list:

    matches = t20_deliveries[
        (t20_deliveries["Bat First"] == team) |
        (t20_deliveries["Bat Second"] == team)
    ]

    match_summary = (
        matches
        .groupby("Match ID")
        .agg(
            Winner=("Winner", "first"),
            Runs=("Runs From Ball", "sum")
        )
    )

    matches_played = len(match_summary)

    wins = (
        match_summary["Winner"] == team
    ).sum()

    losses = (
        (match_summary["Winner"].notna()) &
        (match_summary["Winner"] != team)
    ).sum()

    win_percentage = (
        wins / matches_played * 100
        if matches_played > 0 else 0
    )

    runs_scored = (
        matches[matches["Bat First"] == team]
        .groupby("Match ID")["Innings Runs"]
        .max()
        .sum()
    )

    runs_scored += (
        matches[matches["Bat Second"] == team]
        .groupby("Match ID")["Innings Runs"]
        .max()
        .sum()
    )

    t20_team_stats.append([
        team,
        matches_played,
        wins,
        losses,
        round(win_percentage, 2),
        runs_scored
    ])

t20_team_stats = pd.DataFrame(
    t20_team_stats,
    columns=[
        "Team",
        "Matches",
        "Wins",
        "Losses",
        "Win %",
        "Runs Scored"
    ]
)

t20_team_stats = t20_team_stats.sort_values(
    "Wins",
    ascending=False
)

print("\nTOP T20I TEAMS BY WINS")
print("-" * 40)
print(
    t20_team_stats
    .head(20)
    .to_string(index=False)
)

print("\nTOP T20I TEAMS BY WIN %")
print("-" * 40)

print(
    t20_team_stats[
        t20_team_stats["Matches"] >= 50
    ]
    .sort_values("Win %", ascending=False)
    .head(15)
    .to_string(index=False)
)

print("\nTOP T20I TEAMS BY RUNS SCORED")
print("-" * 40)

print(
    t20_team_stats
    .sort_values("Runs Scored", ascending=False)
    .head(15)
    .to_string(index=False)
)

print("\nTOTAL T20I TEAMS ANALYZED:",
      len(t20_team_stats))

print("\nSTEP 9 COMPLETED")
```

```python
print("=" * 60)
print("STEP 10 - ODI HEAD-TO-HEAD ANALYSIS")
print("=" * 60)

def head_to_head(team1, team2):

    matches = odi_matches[
        (
            (odi_matches["Team 1"] == team1) &
            (odi_matches["Team 2"] == team2)
        ) |
        (
            (odi_matches["Team 1"] == team2) &
            (odi_matches["Team 2"] == team1)
        )
    ].copy()

    if matches.empty:
        print("No matches found between", team1, "and", team2)
        return

    team1_wins = (
        matches["Winner"] == team1
    ).sum()

    team2_wins = (
        matches["Winner"] == team2
    ).sum()

    total_matches = len(matches)

    team1_win_percentage = round(
        team1_wins / total_matches * 100, 2
    )

    team2_win_percentage = round(
        team2_wins / total_matches * 100, 2
    )

    team1_scores = (
        odi_deliveries[
            odi_deliveries["Inning Team"] == team1
        ]
        .groupby("Match ID")["Total Runs"]
        .sum()
    )

    team2_scores = (
        odi_deliveries[
            odi_deliveries["Inning Team"] == team2
        ]
        .groupby("Match ID")["Total Runs"]
        .sum()
    )

    team1_scores = team1_scores[
        team1_scores.index.isin(matches["Match ID"])
    ]

    team2_scores = team2_scores[
        team2_scores.index.isin(matches["Match ID"])
    ]

    print("\nHEAD-TO-HEAD:", team1, "vs", team2)
    print("-" * 40)

    print("Total Matches:", total_matches)

    print(team1, "Wins:", team1_wins)
    print(team2, "Wins:", team2_wins)

    print(team1, "Win %:", team1_win_percentage)
    print(team2, "Win %:", team2_win_percentage)

    print("\n", team1, "Highest Score:",
          team1_scores.max())

    print(team1, "Lowest Score:",
          team1_scores.min())

    print("\n", team2, "Highest Score:",
          team2_scores.max())

    print(team2, "Lowest Score:",
          team2_scores.min())

    history = matches[
        [
            "Match ID",
            "Date",
            "Team 1",
            "Team 2",
            "Winner",
            "Venue",
            "Event Name"
        ]
    ].sort_values("Date")

    print("\nMATCH HISTORY")
    print("-" * 40)
    print(history.tail(20).to_string(index=False))

print("\nExample 1")
head_to_head("India", "Australia")

print("\nExample 2")
head_to_head("India", "Pakistan")
```

```python
def tournament_analysis(event_name):

    matches = odi_matches[
        odi_matches["Event Name"].str.lower() == event_name.lower()
    ].copy()

    if matches.empty:
        print("No exact tournament/event found:", event_name)
        return

    match_ids = matches["Match ID"]

    deliveries = odi_deliveries[
        odi_deliveries["Match ID"].isin(match_ids)
    ].copy()

    teams = pd.unique(
        pd.concat([matches["Team 1"], matches["Team 2"]])
    )

    team_wins = matches["Winner"].value_counts()

    top_batters = (
        deliveries.groupby("Batter")["Runs by Batter"]
        .sum()
        .sort_values(ascending=False)
        .head(10)
    )

    top_bowlers = (
        deliveries[deliveries["Player Out"].notna()]
        .groupby("Bowler")["Player Out"]
        .count()
        .sort_values(ascending=False)
        .head(10)
    )

    individual_scores = (
        deliveries.groupby(["Match ID", "Batter"])["Runs by Batter"]
        .sum()
        .sort_values(ascending=False)
        .head(10)
    )

    print("=" * 60)
    print("TOURNAMENT / EVENT ANALYSIS")
    print("=" * 60)

    print("\nEVENT:", event_name)
    print("TOTAL MATCHES:", len(matches))
    print("TOTAL TEAMS:", len(teams))

    print("\nTEAMS")
    print(", ".join(sorted(teams)))

    print("\nMOST WINS")
    print(team_wins.head(10).to_string())

    print("\nTOP 10 RUN SCORERS")
    print(top_batters.to_string())

    print("\nTOP 10 WICKET TAKERS")
    print(top_bowlers.to_string())

    print("\nTOP 10 HIGHEST INDIVIDUAL SCORES")
    print(individual_scores.to_string())

    print("\nSTEP 11 TOURNAMENT ANALYSIS COMPLETED")
```

```python
def t20_head_to_head(team1, team2):
    matches = t20_deliveries[
        (
            (t20_deliveries["Bat First"] == team1) &
            (t20_deliveries["Bat Second"] == team2)
        ) |
        (
            (t20_deliveries["Bat First"] == team2) &
            (t20_deliveries["Bat Second"] == team1)
        )
    ].copy()

    if matches.empty:
        print("No matches found between", team1, "and", team2)
        return

    match_summary = (
        matches.groupby("Match ID")
        .agg(
            Date=("Date", "first"),
            Bat_First=("Bat First", "first"),
            Bat_Second=("Bat Second", "first"),
            Winner=("Winner", "first"),
            Venue=("Venue", "first")
        )
        .reset_index()
    )

    team1_wins = (match_summary["Winner"] == team1).sum()
    team2_wins = (match_summary["Winner"] == team2).sum()
    total_matches = len(match_summary)

    team1_scores = (
        matches[matches["Bat First"] == team1]
        .groupby("Match ID")["Innings Runs"]
        .max()
    )

    team1_scores = team1_scores[team1_scores.index.isin(match_summary["Match ID"])]

    team2_scores = (
        matches[matches["Bat First"] == team2]
        .groupby("Match ID")["Innings Runs"]
        .max()
    )

    team2_scores = team2_scores[team2_scores.index.isin(match_summary["Match ID"])]

    print("=" * 60)
    print(f"T20I HEAD-TO-HEAD: {team1} vs {team2}")
    print("=" * 60)

    print("Total Matches:", total_matches)
    print(team1, "Wins:", team1_wins)
    print(team2, "Wins:", team2_wins)
    print(team1, "Win %:", round(team1_wins / total_matches * 100, 2))
    print(team2, "Win %:", round(team2_wins / total_matches * 100, 2))

    print(team1, "Highest Score:", team1_scores.max())
    print(team1, "Lowest Score:", team1_scores.min())

    print(team2, "Highest Score:", team2_scores.max())
    print(team2, "Lowest Score:", team2_scores.min())

    print("\nRECENT MATCH HISTORY")

    print(
        match_summary[
            ["Match ID", "Date", "Bat_First", "Bat_Second", "Winner", "Venue"]
        ]
        .sort_values("Date")
        .tail(20)
        .to_string(index=False)
    )


t20_head_to_head("India", "Australia")
t20_head_to_head("India", "Pakistan")
```

```python
def player_search(player_name):

    odi = odi_deliveries[
        odi_deliveries["Batter"].str.contains(player_name, case=False, na=False)
    ]

    t20 = t20_deliveries[
        t20_deliveries["Batter"].str.contains(player_name, case=False, na=False)
    ]

    if odi.empty and t20.empty:
        print("Player not found.")
        return

    if not odi.empty:
        player = odi["Batter"].value_counts().index[0]

        runs = odi["Runs by Batter"].sum()
        matches = odi["Match ID"].nunique()
        fours = (odi["Runs by Batter"] == 4).sum()
        sixes = (odi["Runs by Batter"] == 6).sum()

        match_runs = (
            odi.groupby("Match ID")["Runs by Batter"]
            .sum()
        )

        highest_score = match_runs.max()

        print("=" * 50)
        print("ODI PLAYER PROFILE")
        print("=" * 50)
        print("Player:", player)
        print("Matches:", matches)
        print("Runs:", runs)
        print("Highest Score:", highest_score)
        print("Fours:", fours)
        print("Sixes:", sixes)

    if not t20.empty:
        player = t20["Batter"].value_counts().index[0]

        runs = t20["Batter Runs"].sum()
        matches = t20["Match ID"].nunique()
        fours = (t20["Batter Runs"] == 4).sum()
        sixes = (t20["Batter Runs"] == 6).sum()

        match_runs = (
            t20.groupby("Match ID")["Batter Runs"]
            .sum()
        )

        highest_score = match_runs.max()

        print("\n" + "=" * 50)
        print("T20I PLAYER PROFILE")
        print("=" * 50)
        print("Player:", player)
        print("Matches:", matches)
        print("Runs:", runs)
        print("Highest Score:", highest_score)
        print("Fours:", fours)
        print("Sixes:", sixes)


player_search("V Kohli")
```

```python
def team_search(team_name):

    odi_matches_team = odi_matches[
        (odi_matches["Team 1"].str.contains(team_name, case=False, na=False)) |
        (odi_matches["Team 2"].str.contains(team_name, case=False, na=False))
    ].copy()

    t20_matches_team = (
        t20_deliveries[
            (t20_deliveries["Bat First"].str.contains(team_name, case=False, na=False)) |
            (t20_deliveries["Bat Second"].str.contains(team_name, case=False, na=False))
        ][["Match ID", "Winner"]]
        .drop_duplicates()
    )

    if odi_matches_team.empty and t20_matches_team.empty:
        print("Team not found.")
        return

    if not odi_matches_team.empty:
        team = (
            pd.concat([
                odi_matches_team["Team 1"],
                odi_matches_team["Team 2"]
            ])
            .value_counts()
            .index[0]
        )

        matches = len(odi_matches_team)
        wins = (odi_matches_team["Winner"] == team).sum()
        losses = (
            (odi_matches_team["Winner"].notna()) &
            (odi_matches_team["Winner"] != team)
        ).sum()

        win_percentage = round(wins / matches * 100, 2)

        runs = odi_deliveries[
            odi_deliveries["Inning Team"] == team
        ]["Total Runs"].sum()

        print("=" * 50)
        print("ODI TEAM PROFILE")
        print("=" * 50)
        print("Team:", team)
        print("Matches:", matches)
        print("Wins:", wins)
        print("Losses:", losses)
        print("Win Percentage:", win_percentage)
        print("Runs Scored:", runs)

    if not t20_matches_team.empty:
        team_matches = t20_deliveries[
            (t20_deliveries["Bat First"] == team) |
            (t20_deliveries["Bat Second"] == team)
        ][["Match ID", "Winner"]].drop_duplicates()

        matches = len(team_matches)
        wins = (team_matches["Winner"] == team).sum()
        losses = (
            (team_matches["Winner"].notna()) &
            (team_matches["Winner"] != team)
        ).sum()

        win_percentage = round(wins / matches * 100, 2)

        runs = t20_deliveries[
            (t20_deliveries["Bat First"] == team) |
            (t20_deliveries["Bat Second"] == team)
        ]["Innings Runs"].groupby(
            t20_deliveries[
                (t20_deliveries["Bat First"] == team) |
                (t20_deliveries["Bat Second"] == team)
            ]["Match ID"]
        ).max().sum()

        print("\n" + "=" * 50)
        print("T20I TEAM PROFILE")
        print("=" * 50)
        print("Team:", team)
        print("Matches:", matches)
        print("Wins:", wins)
        print("Losses:", losses)
        print("Win Percentage:", win_percentage)
        print("Runs Scored:", runs)


team_search("India")
```

```python
def tournament_search(event_name):

    matches = odi_matches[
        odi_matches["Event Name"].str.contains(
            event_name, case=False, na=False
        )
    ].copy()

    if matches.empty:
        print("Tournament/Event not found.")
        return

    match_ids = matches["Match ID"]

    deliveries = odi_deliveries[
        odi_deliveries["Match ID"].isin(match_ids)
    ].copy()

    teams = pd.unique(
        pd.concat([matches["Team 1"], matches["Team 2"]])
    )

    team_wins = (
        matches["Winner"]
        .value_counts()
        .head(10)
    )

    top_batters = (
        deliveries.groupby("Batter")["Runs by Batter"]
        .sum()
        .sort_values(ascending=False)
        .head(10)
    )

    top_bowlers = (
        deliveries[deliveries["Player Out"].notna()]
        .groupby("Bowler")["Player Out"]
        .count()
        .sort_values(ascending=False)
        .head(10)
    )

    highest_scores = (
        deliveries.groupby(
            ["Match ID", "Batter"]
        )["Runs by Batter"]
        .sum()
        .sort_values(ascending=False)
        .head(10)
    )

    print("=" * 60)
    print("TOURNAMENT / EVENT PROFILE")
    print("=" * 60)
    print("Event:", event_name)
    print("Total Matches:", len(matches))
    print("Total Teams:", len(teams))

    print("\nTEAMS")
    print(", ".join(teams))

    print("\nMOST WINS")
    print(team_wins.to_string())

    print("\nTOP 10 RUN SCORERS")
    print(top_batters.to_string())

    print("\nTOP 10 WICKET TAKERS")
    print(top_bowlers.to_string())

    print("\nTOP 10 HIGHEST INDIVIDUAL SCORES")
    print(highest_scores.to_string())


tournament_search("ICC Cricket World Cup")
```

```python
world_cup_matches = odi_matches[
    odi_matches["Event Name"].str.contains(
        "World Cup", case=False, na=False
    )
].copy()

team1 = world_cup_matches[
    ["Match ID", "Team 1", "Winner"]
].rename(columns={"Team 1": "Team"})

team2 = world_cup_matches[
    ["Match ID", "Team 2", "Winner"]
].rename(columns={"Team 2": "Team"})

wc_teams = pd.concat(
    [team1, team2],
    ignore_index=True
)

wc_team_stats = wc_teams.groupby("Team").agg(
    Matches=("Match ID", "nunique"),
    Wins=("Winner", lambda x: (
        x == wc_teams.loc[x.index, "Team"]
    ).sum())
)

wc_team_stats["Losses"] = (
    wc_team_stats["Matches"] -
    wc_team_stats["Wins"]
)

wc_team_stats["Win Percentage"] = (
    wc_team_stats["Wins"] /
    wc_team_stats["Matches"] * 100
).round(2)

wc_team_stats = wc_team_stats.sort_values(
    "Win Percentage",
    ascending=False
)

top_10_wc_teams = (
    wc_team_stats["Win Percentage"]
    .head(10)
    .sort_values()
)

plt.figure(figsize=(10, 6))
top_10_wc_teams.plot(kind="barh")
plt.title("Top 10 World Cup Teams by Win Percentage")
plt.xlabel("Win Percentage")
plt.ylabel("Team")
plt.tight_layout()
plt.show()
```
